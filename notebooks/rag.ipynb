```python
import pandas as pd
```

```python
sentiment_df = pd.read_csv('../data/player_scraped_info.csv')
stats_df = pd.read_csv('../data/player_stats.csv')
```

```python
combined_df = stats_df.merge(
    sentiment_df, 
    on=['playerId', 'name'],  
    how='left'  
)
```

```python
def create_docs(row):
    """
    Create detailed natural language documents using only the specified columns
    """
    
    
    doc_text = f"""
        PLAYER PROFILE: {row['name']} (ID: {row.get('playerId', 'Unknown')})

        BASIC INFORMATION:
        - Position: {row['position']}
        - Team: {row.get('proTeam', 'Unknown')}
        - Position Rank: {row.get('posRank', 'Unranked')}
        - Current Lineup Slot: {row.get('lineupSlot', 'Not set')}
        - Acquisition Method: {row.get('acquisitionType', 'Unknown')}
        - Team Assignment: Team #{row.get('onTeamId', 'Unknown')}

        ELIGIBLE POSITIONS:
        - Can play in slots: {row.get('eligibleSlots', 'No data available')}

        HEALTH STATUS:
        - Injury Status: {row.get('injuryStatus', 'Unknown')}
        - Currently Injured: {'Yes' if row.get('injured', False) else 'No'}

        FANTASY PERFORMANCE:
        - Total Points This Season: {row.get('total_points', 0)} points
        - Average Points Per Game: {row.get('avg_points', 0)} points
        - Projected Season Total: {row.get('projected_total_points', 0)} points
        - Projected Weekly Average: {row.get('projected_avg_points', 0)} points

        OWNERSHIP METRICS:
        - League Ownership: {row.get('percent_owned', 0)}% of leagues own this player
        - Start Rate: {row.get('percent_started', 0)}% of owners start this player
        - Popularity Assessment: {'Widely owned popular player' if row.get('percent_owned', 0) > 70 else 'Moderately owned' if row.get('percent_owned', 0) > 30 else 'Lightly owned sleeper candidate'}

        Player Sentiment SUMMARY:
        {row.get('sentiment', 'No data available')}

        

    """.strip()
            

        
    return doc_text
```

```python
combined_df['stats_doc'] = combined_df.apply(create_docs, axis=1)

combined_df.head()
```

```python
cols_to_drop = ['stats', 'reddit_text', 'fantasy_pros_text', 'espn_text', 'posRank', 'eligibleSlots', 'lineupSlot',
       'acquisitionType', 'proTeam', 'onTeamId', 'position', 'injuryStatus',
       'injured', 'total_points', 'avg_points', 'projected_total_points', 'projected_avg_points', 'percent_owned', 'percent_started']

existing_cols = [col for col in cols_to_drop if col in combined_df.columns]
if existing_cols:
    combined_df.drop(columns=existing_cols, inplace=True)
else:
    print("None of the specified columns to drop are present in the DataFrame.")
combined_df.columns
combined_df.head()
```

```python
import getpass
import os

if not os.environ.get("OPENAI_API_KEY"):
  os.environ["OPENAI_API_KEY"] = getpass.getpass("Enter API key for OpenAI: ")

from langchain.chat_models import init_chat_model

llm = init_chat_model("gpt-4o-mini", model_provider="openai")
```

```python
from langchain_openai import OpenAIEmbeddings

embeddings = OpenAIEmbeddings(model="text-embedding-3-large")
```

```python
from langchain_chroma import Chroma

vector_store = Chroma(
    collection_name="example_collection",
    embedding_function=embeddings,
    persist_directory="./chroma_langchain_db",  
)
```

```python
from langchain_community.document_loaders.csv_loader import CSVLoader
import numpy as np

loader = CSVLoader(file_path="rag_docs.csv")

docs = loader.load()
np.mean([len(doc.page_content) for doc in docs])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=2000,  
    chunk_overlap=400,  # chunk overlap (characters)
    add_start_index=True,  # track index in original document
)
all_splits = text_splitter.split_documents(docs)

print(f"Split blog post into {len(all_splits)} sub-documents.")
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template("""
You are an expert fantasy football advisor with deep knowledge of player performance, injuries, matchups, and league trends.

Use the following player information to provide helpful fantasy football advice. Be specific about why you recommend or don't recommend a player.

IMPORTANT GUIDELINES:
- ONLY reference these specific stats in your recommendations: Total Points, Average Points Per Game, Projected Season Total, Projected Weekly Average, League Ownership %, Start Rate %
- Use all other information (injury status, team, position, expert analysis, etc.) as CONTEXT to inform your advice, but do not quote specific numbers from those areas
- Focus on the measurable fantasy performance data listed above for concrete recommendations

PLAYER CONTEXT:
{context}

QUESTION: {question}

FANTASY ADVICE:
- Give clear start/sit recommendations when relevant
- Base recommendations primarily on the allowed stats: points, projections, and ownership rates
- Explain your reasoning using the provided performance data
- Use injury status, expert opinions, and other context to support your analysis (without citing specific details)
- If you don't have enough statistical information, say so clearly
- Keep advice practical and actionable

Answer:
""")
```

```python
from langchain_core.documents import Document
from typing_extensions import List, TypedDict


class State(TypedDict):
    question: str
    context: List[Document]
    answer: str

```

```python
def retrieve(state: State):
    retrieved_docs = vector_store.similarity_search(state["question"])
    return {"context": retrieved_docs}


def generate(state: State):
    docs_content = "\n\n".join(doc.page_content for doc in state["context"])
    messages = prompt.invoke({"question": state["question"], "context": docs_content})
    response = llm.invoke(messages)
    return {"answer": response.content}
```

```python
from langgraph.graph import START, StateGraph

graph_builder = StateGraph(State).add_sequence([retrieve, generate])
graph_builder.add_edge(START, "retrieve")
graph = graph_builder.compile()
```

```python
result = graph.invoke({"question": "Who is one player I should try to draft"})
```
